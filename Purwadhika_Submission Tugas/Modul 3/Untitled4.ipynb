```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from string import punctuation
```

```python
text = "Halo, selamat malam. Nama Saya Andre. Saat ini saya sedang makan mie ayam harganya rp 25.000. Anda mau?"
text
```

## Normalization

```python
text = text.lower()
text
```

## Tokenize

```python
sent_tokenize(text)
```

```python
word_tokenize(text)
```

```python
tokens = word_tokenize(text)
```

```python
tokens
```

## Punctuation Removal

```python
punctuation
```

```python
result = []
for i in tokens:
    if i not in punctuation:
        result.append(i)
result
```

```python
result2 = [i for i in tokens if i not in punctuation]
```

```python
result2
```

## Alpha Numeric Cleansing

```python
tokens2 = [i for i in tokens if i.isalnum()]
```

```python
tokens2
```

## Stopwords Removal

```python
stopwords.words('indonesian')
```

```python
tokens3 = [i for i in tokens2 if i not in stopwords.words('indonesian')]
```

```python
tokens3
```
