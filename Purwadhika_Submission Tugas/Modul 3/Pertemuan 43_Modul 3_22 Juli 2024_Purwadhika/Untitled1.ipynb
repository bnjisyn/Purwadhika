```python
### Power Transformer
- Salah satu alternatif untuk menghadle data yg memiliki Distribusi Skew (dengan atau tanpa Outliers)
- Akan di transformasikan menjadi terdistribusi normal atau mendekati Normal
- Metode yg digunakan ada 
    - Yeo-johnson : Bisa digunakan untuk yg SKew positif maupun negatif
    - Box cox : Lebih cocok digunakan untuk yg memiliki Skew Positif

## Binary Encoder
- Salah satu alternatif untuk proses Encoding Kolom yg bertipe data Nominal (tanpa urutan)
- Jumlah Data unik akan dikonversikan sesuai bilangan Biner, misal Jumlah Data unik ada 15, akan dikonversi menjadi 4 Kolom baru
- Kolom lebih sulit diinterpretasikan jika dibandingkan dengan metode One Hot Encoding, tetapi jauh lebih efisien (jumlah kolom baru yg terbentuk Jauh lebih sedikit)

## Scaling
- Untuk menyamakan skalan nilai dari Features
- Untuk Features yg berisi data Continous Numerik
- Jika Ada perbedaan Nilai yg signifikan antar Features
    - Contoh : Gaji - skala Jutaan, Jam tidur - skala satuan
- Tidak menggunakan Algoritma berbasis Pohon (Tree based)

## Jenis Scaling
## Standar Scaller :
- Data harus terdistribusi Normal
- Proses nya : Mengubah Nilai Mean menjadi 0
- Kemudian mengubah nilai Standar Deviasi menjadi 1

## MinMax Scaller :
- Mengubah Range Nilai menjadi 0 sampai 1
- Proses nya : Mengubah nilai minimun menjadi 0
- Kemudian mengubah nilai Maksimum menjadi 1

## Robust Scaller :
- Mengubah nilai median (Q2) menjadi 0
- Kemudian mengubah nilai IQR (selisih antara Q3 dan Q1) menjadi 1
- efektif jika ada Outliers
- Bisa digunakan untuk hampir semua jenis keadaan data
```
