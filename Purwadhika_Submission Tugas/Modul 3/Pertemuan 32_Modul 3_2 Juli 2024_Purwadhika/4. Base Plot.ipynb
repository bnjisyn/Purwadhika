```python
x = np.arange(1, 11)
y = x ** 2
z = x ** 3
```

```python
plt.figure(figsize=(8, 7))
plt.plot(x,x, x,y, x,z)
plt.show()
```

```python
plt.figure(figsize=(8, 7))
plt.plot(x,x, 'red', x,y, 'blue', x,z, 'black')
plt.title("Latihan Basic")
plt.xlabel("Nilai X")
plt.ylabel("Nilai X/Y/Z")
plt.grid(True)
plt.legend(["Garis XX", "Garis XY", "Garis XZ"])
plt.show()
```

```python
plt.figure(figsize=(8, 7))
plt.plot(x,x, 'r--', x,y, 'b:', x,z, 'k-.')
plt.title("Latihan Basic")
plt.xlabel("Nilai X")
plt.ylabel("Nilai X/Y/Z")
plt.grid(True)
plt.legend(["Garis XX", "Garis XY", "Garis XZ"])
plt.show()
```

```python
plt.figure(figsize=(8, 7))
plt.plot(x,y,
         color = 'b',
        marker='v',
        markerfacecolor = 'g',
        markersize=21)
plt.title("Latihan Basic")
plt.xlabel("Nilai X")
plt.ylabel("Nilai Y")
plt.grid(True)
plt.legend(["Garis XY"])
plt.show()
```

```python
plt.figure(figsize=(8, 7))
plt.plot(x,y,
         color = 'b',
        marker='v',
        markerfacecolor = 'g',
        markersize=21)
plt.title("Latihan Basic")
plt.xlabel("Nilai X")
plt.ylabel("Nilai Y")
plt.grid(True)
plt.legend(["Garis XY"])
plt.text(8, 20, "Cek Lokasi") ## Memunculkan Teks di posisi tertentu
# plt.text (SumbuX, sumbu Y, Text yg dimunculkan)
for i in y:
    ind = list(y).index(i)
    X = x[ind]
    plt.text(X-0.1, i+4, i)
plt.show()
```
