## Elbow Analysis
- Digunakan untuk mencari/mendapatkan jumlah cluster Ideal
- Inertia ==> Jumlah pangkat (sum of squared) Jarak data point ke Centroid terdekat
- Semakin kecil semakin Baik
- Julah cluster (K teroptimal) adalah nilai K ketika penurunan Inertia sudah tidak signifikan


```python
score = []
for i in range(2, 21): ##Jumlah cluster yg ingin kita cobain
    model = KMeans(n_clusters=i, random_state=25)
    model.fit_predict(X)
    score.append(model.inertia_)
```

```python
plt.figure(figsize=(13, 8))
plt.plot(np.arange(2, 21), score, 'r-')
plt.plot(np.arange(2, 21), score, 'go')
plt.xlabel('Jumlah Cluster')
plt.ylabel('Inertia')
plt.show()
```

```python
## Silhouette Score
- Semakin tinggi semakin baik
- Paling tinggi => 1
```

```python
from sklearn.metrics import silhouette_score
```

```python
score_s = []

for i in range(2, 21):
    model_K = KMeans(n_clusters=i, random_state=25)
    cluster = model_K.fit_predict(X)
    score_s.append(silhouette_score(X, cluster))
```

```python
plt.figure(figsize=(13, 8))
plt.plot(np.arange(2, 21), score_s, 'g-')
plt.plot(np.arange(2, 21), score_s, 'ro')
plt.xlabel('Jumlah Cluster')
plt.ylabel('Silhouette Score')
plt.show()
```
