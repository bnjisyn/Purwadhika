```python
## Duplicate / Redundant Data Handling
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({
'Brand' :['Sedaap','Sedaap', 'Indomie','Indomie','Indomie'],
'Package':['Cup', 'Cup', 'Cup', 'Pack', 'Pack'],
'Rating' :[4, 4, 2.5, 4.5, 5]})
df
```

```python
df.drop_duplicates() ## Akan menghapus baris yg memiliki duplikat, Isi dari semua kolom Sama
## Umumnya pake ini tetapi bisa d sesuaikan dg kebutuhan
```

```python
df
```

```python
df.drop_duplicates(subset=['Brand']) ## Hapus baris duplikat berdasarkan kolom tertentu (Brand)
```

```python
df.drop_duplicates(subset=['Brand','Package'])
## Menggunakan nilai Keep default ==> keep ==> First
```

```python
df.drop_duplicates(subset=['Brand','Package'], keep = 'last')
```
