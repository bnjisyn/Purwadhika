```python
plt.figure(figsize=(8, 7)) ## mengatur ukuran plot (Horizontal, vertikal)
plt.plot(x,y, x,z) ## Membuat plot
plt.title("Latihan Data Vis", size=25) ## membuat judul dan ukuran font judul Plot
plt.xlabel("Sumbu X", size=20) ## Membuat Label pada sumbu X dan ukuran font nya
plt.ylabel("Sumbu Y", size=20) ## Membuat Label pada sumbu Y dan ukuran font nya
plt.grid(True) ## menampilkan grid
plt.legend(['Garis XY', 'Garis XZ'], loc=0) ## menampilkan Legend, loc = Posisi Legend
# Loc  = 0, Best Position menurut sistem, secara default Loc akan dibuat sesuai Best Positionnya (nilai 0)
## Setup Loc manual ==> 1 - 10
plt.savefig('PlotLatihan.png')
plt.show() ## menampilkan plot
```
