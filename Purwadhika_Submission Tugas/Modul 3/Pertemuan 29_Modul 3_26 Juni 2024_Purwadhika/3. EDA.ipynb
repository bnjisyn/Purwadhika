## 1. Business Question : Sub Category apa yg memiliki Penjualan (items) Tertinggi?

- **Insight** - Informasi : Penjualan tertinggi (items) dari Sub category detergent sebanyak 36.000 transaksi

## 2. Business Question : Sub Category apa yg memiliki penjualan (items) paling sedikit?

```python
df['sub_category'].value_counts()
```

```python
pd.crosstab(df['sub_category'], columns='Jumlah').sort_values('Jumlah')
```

- **Insight** - Informasi : Penjualan paling sedikit (items) dari sub category Rice sebanyak 12.000 transaksi

## 3. Business Question : Lokasi (format) mana yg memiliki penjualan paling sedikit?

```python
df['format'].value_counts()
```

- Insight - Informasi :
    - Penjualan paling sedikit (items) dari lokasi/format Hypermarket sebanyak 5371
    - Penjualan paling tinggi ada di Minimarket sebanyak 46.803, karena lokasi minimarket dekat dg perumahan dan jumlahnya cukup banyak. rata-rata jarakl 100m dari perumahan, jumlah minimal ada 2 minimarket
    
source : https://bisnis.co.id/rata-rata-jarak-minimarket

```python
pd.crosstab(df['format'], columns='Jumlah').sort_values('Jumlah')
```

## Insight Data - Informasi :
- Transaksi paling tinggi ada di minimarket, sedangkan paling rendah ada di hypermarket
- Jika akan merilis produk baru, sebaikonya dilakukan di minimarket
- Jika akan melakukan promo produk sebaiknya di minimarket
- Penjualan di minimarket tinggi karena lokasi minimarket dekat dg perumahan dan jumlahnya cukup banyak. rata-rata jarakl 100m dari perumahan, jumlah minimal ada 2 minimarket ==> Hipotesa/asumsi
- Jika ada Hipotesa - dugaan - asumsi harus berdasarkan data
