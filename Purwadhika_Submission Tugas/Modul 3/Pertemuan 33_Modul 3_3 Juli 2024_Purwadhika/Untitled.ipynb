```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df_tips = sns.load_dataset('tips')
df_tips.head()
```

```python
## Histogram - Melihat distribusi data numerik - continous
```

```python
## matplotlib
plt.hist(df_tips['total_bill'])
plt.show()
```

```python
## matplotlib
plt.figure(figsize=(8,7))
plt.hist(df_tips['total_bill'], bins=15, color='green')
plt.title('Histogram dari Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.grid(True)
plt.legend(['Total Bill Hist'])
plt.savefig('Hist_Bill.png')
plt.show()
```

```python
## Seaborn
#plt.figure(figsize=(8,7))
sns.displot(df_tips['tip'], bins=15)
plt.title('Tip Histogram')
plt.show()
```

```python
## Pandas
df_tips['tip'].hist(bins=15, color='green', figsize=(8,7))
plt.title('Tip Histogram')
plt.xlabel('Tip')
plt.ylabel("freq")
plt.show()
```

```python
## Boxplot ==> Melihat distribusi data dan Descr Stats, juga untuk melihat / Cek Outliers ==> Kolom Numerik COntinous
```

```python
## Matplotlib
plt.boxplot(df_tips['total_bill'])
plt.title("Box plot Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Value")
plt.show()
```

```python
## Matplotlib
plt.figure(figsize=(8,7))
plt.boxplot(df_tips['total_bill'])
plt.title("Box plot Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Value")
plt.grid(True)
plt.savefig('Box3.png')
plt.show()
```

```python
## LinePlot ==> Untuk melihat Tren ==> berkaitan dg Waktu / Time Based
```

```python
set(sns.get_dataset_names())
```

```python
df_air = sns.load_dataset('flights')
df_air.head()
```

```python
df_Mar = df_air[df_air['month'] == 'Mar']
df_Mar
```

```python
## Matplotlib
plt.figure(figsize=(8,7))
plt.plot(df_Mar['year'], df_Mar['passengers'], 'red')
plt.title("Passengers Trends on Mar (Yearly)")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

```python
### Seaborn
sns.lineplot(data = df_air, x = 'year', y='passengers')
plt.title("Tren Rata-rata passengers per Tahun")
plt.show()
```

```python
### Seaborn
sns.lineplot(data = df_air, x = 'year', y='passengers', hue='month')
plt.title("Tren Rata-rata passengers per Tahun")
plt.show()
```

```python
## Pandas
df_Mar.plot.line(x='year', y='passengers')
plt.title("Passengers Trends on Mar Yearly")
plt.show()
```
