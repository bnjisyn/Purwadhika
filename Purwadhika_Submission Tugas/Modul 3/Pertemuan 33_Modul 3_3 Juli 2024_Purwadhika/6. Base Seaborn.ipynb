```python
## Barplot
```

```python
sns.barplot(x='sex', y='total_bill', data=df_tips)
plt.title("Barchart dari Seaborn")
plt.show()
## Menggunakan Fungsi aggregasi Mean
```

```python
sns.barplot(x='sex', y='total_bill', data=df_tips, errorbar=None)
plt.title("Barchart dari Seaborn")
plt.show()
## Menggunakan Fungsi aggregasi Mean
```

```python
sns.barplot(x=df_tips['sex'], y=df_tips['total_bill'], errorbar=None)
plt.title("Barchart dari Seaborn")
plt.show()
## Menggunakan Fungsi aggregasi Mean
```
