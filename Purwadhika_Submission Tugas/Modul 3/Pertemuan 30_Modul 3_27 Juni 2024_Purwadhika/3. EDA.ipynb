## Analisa Univariate menggunakan Crosstab
- Kolom yg dianalisa menggunakan crosstab (secara univariate) adalah Kolom Kategorikal (datanya kategorik)
- Kolom yg berisi data numerik continous bisa dilakukan tetapi tidak direkomendasikan, karena akan sulit mengambil insight data/informasi
- Fungsi Aggregate yg digunakan di Crosstab secara default adalah fungsi **Count**
- Output yg dihasilkan (ketika analisa univariate) sama dengan ketika menggunakan fungsi **.value_counts()**
- Perbedaan terletak pada tipe data yg dihasilkan. Crosstab menghasilkan Dataframe, value_counts menghasilkan Series
