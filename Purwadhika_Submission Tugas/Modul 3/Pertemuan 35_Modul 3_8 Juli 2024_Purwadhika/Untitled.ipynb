```python
## Simple Linear Regression
Regresi dg SATU Independent Variable
Y = aX + b
Y => Dependent variable
X => Independetn Variable
a => Koefisien
b => Konstanta - Intercept
```

```python
data = {
"mesin" : [2000, 5000, 2500, 7000, 4000, 3500],
"harga" : [500, 1500, 700, 1800, 1000, 800]}
```

```python
df = pd.DataFrame(data).sort_values('mesin')
df
```

```python
plt.plot(df['mesin'], df['harga'], 'r*-')
plt.xlabel('Mesin')
plt.ylabel('Harga')
plt.show()
```

```python
from sklearn import linear_model
```

```python
.fit() ==> Model belajar ==> Latihan ==> Training
==> Model belajar mencari PATTERN atau POLA dari data historikal (data yg sudah ada)


==> Kita memberikan Soal & Jawaban untuk Model belajar
Soal ==> Kolom Features ==> X ==> Independent Variable
Jawaban ==> Kolom target/Label ==> y ==> Dependent Variable

.fit(X, y)

====> Memberikan ujian - Test
.predict() ==> Model melakukan Ujian ==> Menjawab soal test yg kita berikan
Soal Test ==> Data baru yg formatnya sesuai dg Soal Latihan (X)

.predict(Soal Ujian)

===> Proses penilaian Model mengerjakan Ujian/Test ==> Evaluation Metrics
```

```python
Model = linear_model.LinearRegression()
```

```python
Linear Regression ==> Salah satu Algoritma yg digunakan untuk kasus Supervised Learning Regression
```

```python
### Tentukan Soal (Kolom features - X) dan Jawaban (Kolom target - y)
X = df[['mesin']] ## Kolom X selalu gunakan Format Dataframe
y = df['harga'] ## Gunakan Series
```

```python
## Model belajar dari Soal & Jawaban untuk Mencari Pattern
Model.fit(X, y)
```

```python
## Model mengerjakan Ujian / Test dg Data baru / Melakukan Prediksi terhadap data baru
## Data Baru ==> Data yg tidak ada ketika Model belajar
```

```python
Y = ax + b

Y = Harga rumah
x = Jumlah kamar
b = Konstanta / Harga default

Y = 500x + 2000

Y = Dependent Var ==> Harga rumah
X = Independent Variabel => Jumlah kamar

b = Konstanta = Intercept = Harga Default = 2000
Adalah harga ketika nilai X (Independent Var) = 0
Jika jumlah kamar = 0, Harga rumah = 2000

a = Koefisien = slope = Gradient = Menampilkan seberapa besar pengaruh Independent Variabel (X) terhadap dependent Var (Y)
=> Mernampilkan seberqapa besar pengaruh Jumlah kamar terhadap Harga rumah

x = 1, Jumlah kamar = 1

Harga rumah akan bertambah sebesar a * x (500 * 1) = 500
```
