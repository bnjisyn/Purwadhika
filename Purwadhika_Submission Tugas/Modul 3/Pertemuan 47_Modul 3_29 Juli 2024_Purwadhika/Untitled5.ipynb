```python
import numpy as np
import pandas as pd
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('movie_dataset.csv')
df.head()
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
fitur = ["genres", "cast", "director"]
```

```python
for i in fitur:
    df[i] = df[i].fillna("")
```

```python
df.isna().sum()
```

```python
def combo(x):
    return x['genres']  + " " + x["cast"] + " " + x["director"]
```

```python
df["combo_features"] = df.apply(combo, axis=1)
```

```python
df.head()
```

```python
## Recommendation System
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
CV = CountVectorizer()
```

```python
movie_matrix = CV.fit_transform(df['combo_features'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cos_score = cosine_similarity(movie_matrix)
```

```python
## cos_score ini lah model kita
```

```python
## Contoh Recommendation
```

```python
## Index by Movie
```

```python
def index_by_movie(movie):
    
    return list(df[df['title'] == movie]['index'])[0]
```

```python
## Movie by Index
```

```python
def movie_by_index(index):
    return list(df[df['index'] == index]['title'])[0]
```

```python
df[['title', 'original_title']]
```

```python
watched_movie = "Spectre"
```

```python
index_watched = index_by_movie(watched_movie)
```

```python
index_watched
```

```python
similar_movie 
```

```python
similar_movie = list(enumerate(cos_score[index_watched]))
```

```python
similar_movie[:6]
```

```python
sorted_movies = sorted(similar_movie, key=lambda x: x[1], reverse=True) [1:]
```

```python
print(f"Top 5 Movies Recommendation Based on your Watched Movie : {watched_movie}")
print("=" * 65)
count = 1
for i in sorted_movies:
    print(movie_by_index(i[0]))
    count += 1
    if count == 6:
        break
```
